## 1. Introduction to NumPy

### 1.1 What Is NumPy?

**NumPy** (*Numerical Python*) adalah pustaka Python yang dirancang untuk komputasi numerik dan ilmiah.
NumPy memudahkan pengorganisasian data numerik, pelaksanaan operasi pada seluruh array sekaligus, serta persiapan data untuk analisis statistik, pemodelan ilmiah, dan *machine learning*.

### 1.3 Python Lists vs. NumPy Arrays

**Python** List dapat menyimpan berbagai tipe data.

**NumPy Array** menyimpan elemen dengan tipe data yang sama yang lebih efisien untuk perhitungan numerik. 

Misal operator `+` pada List untuk menggabungkan elemen, pada NumPy Array untuk penjumlahann elemen yang bersesuaian.

```python
python_list = [2, 4, 6]
numpy_array = np.array([2, 4, 6])

print("List Python:", python_list)
print("Array NumPy:", numpy_array)
print("Tipe Python:", type(python_list))
print("Tipe NumPy:", type(numpy_array))
```

```python
print(python_list + python_list)
print(numpy_array + numpy_array)
```

## 2. Creating NumPy Arrays

### 2.1 Creating Arrays from Python Lists

Fungsi `np.array()` digunakan untuk mengubah data Python menjadi NumPy Array. Jumlah dimensi ditentukan oleh struktur data yang dimasukkan.
- **Array 1D:** Berbentuk vektor.
- **Array 2D:** Berbentuk matriks (baris dan kolom).
- **Array 3D:** Kumpulan matriks 2D.
- **Array 4D:** Kumpulan array 3D.

Harus memiliki bentuk (`shape`) yang konsisten. **Shape:** Ukuran NumPy Array yang menunjukkan jumlah elemen pada setiap dimensi.

### 2.2 Creating a One-Dimensional Array (1D)

Array 1D berguna untuk merepresentasikan pengukuran dari **satu variabel**, misalnya nilai lima orang mahasiswa.

```python
student_scores = [75, 80, 92, 86, 88]
scores_1d = np.array(student_scores)

print(scores_1d)
print("Dimensi:", scores_1d.ndim)
print("Bentuk (shape):", scores_1d.shape)
```

### 2.3 Creating a Two-Dimensional Array (2D)

Array 2D merepresentasikan data **dalam baris dan kolom**. Pada contoh di bawah, setiap baris mewakili seorang mahasiswa, sedangkan setiap kolom mewakili penilaian yang berbeda.

```python
student_assessments = [[80, 85, 90],
                       [75, 88, 84],
                       [92, 90, 95]]

scores_2d = np.array(student_assessments)

print(scores_2d)
print("Dimensi:", scores_2d.ndim)
print("Bentuk (shape):", scores_2d.shape)
```

Bentuk `(3, 3)` menunjukkan **tiga baris dan tiga kolom**.

### 2.4 Creating a Three-Dimensional Array (3D)

Array 3D dapat dipahami sebagai **tumpukan matriks**.

```python
week_1 = np.array([[10, 12],
                   [14, 16]])
week_2 = np.array([[11, 13],
                   [15, 17]])

observations_3d = np.array([week_1, week_2])

print(observations_3d)
print("Dimensi:", observations_3d.ndim)
print("Bentuk (shape):", observations_3d.shape)
```

Bentuk `(2, 2, 2)` menunjukkan **dua matriks**, masing-masing terdiri dari **dua baris dan dua kolom**. Sumbu pertama menentukan matriks mana yang dipilih.

### 2.5 Creating a Four-Dimensional Array (4D)

Array 4D menambahkan satu tingkat pengorganisasian lagi. Anda dapat membayangkannya sebagai kumpulan array 3D. Struktur seperti ini muncul pada data ilmiah dan kumpulan (*batch*) pengamatan multidimensi.

```python
first_cube = np.array([[[1, 2], [3, 4]],
                       [[5, 6], [7, 8]]])
second_cube = np.array([[[11, 12], [13, 14]],
                        [[15, 16], [17, 18]]])

observations_4d = np.array([first_cube, second_cube])

print(observations_4d)
print("Dimensi:", observations_4d.ndim)
print("Bentuk (shape):", observations_4d.shape)
```

### 2.6 Creating Arrays from Scratch: `np.zeros()`
`np.zeros()` langsung membuat array yang berisi nol.

```python
zero_vector = np.zeros(5)
zero_matrix = np.zeros((3, 4))

print("Nol 1D:")
print(zero_vector)
print("Nol 2D:")
print(zero_matrix)
```

### 2.7 Creating Random Arrays: `np.random.random()`

`np.random.random()` menghasilkan bilangan desimal acak antara **0 hingga kurang dari 1**. Penggunaan `seed` membuat hasil bilangan acak tetap sama setiap kali kode dijalankan.

```python
np.random.seed(42)
random_measurements = np.random.random((2, 3))

print(random_measurements)
print("Bentuk (shape):", random_measurements.shape)
```

### 2.8 Creating Sequences: `np.arange()`

`np.arange(start, stop, step)` digunakan untuk membuat deret angka dengan nilai awal (`start`), batas akhir (`stop`), dan jarak antarangka (`step`). Nilai `stop` tidak disertakan. Secara default, `start = 0` dan `step = 1`.

```python
sequence_a = np.arange(5)
sequence_b = np.arange(2, 11, 2)
sequence_c = np.arange(-3, 4, 3)

print("Deret default:", sequence_a)
print("Deret bilangan genap:", sequence_b)
print("Deret dengan langkah:", sequence_c)
```

## 3. Array Attributes and Data Types

### 3.1 Array Dimensionality and Axes

**Axis** adalah sumbu yang menunjukkan arah dimensi pada NumPy Array. Pada array 2D, `axis=0` menunjukkan arah baris dan `axis=1` menunjukkan arah kolom. Pada array 3D, terdapat tiga sumbu, yaitu lapisan, baris, dan kolom.

- **`.ndim`**: Jumlah dimensi atau sumbu.
- **`.shape`**: Ukuran setiap dimensi.
- **`.size`**: Jumlah seluruh elemen array.

| Atribut | Arti | Contoh hasil |
|---|---|---|
| `.ndim` | Jumlah dimensi/sumbu | `2` |
| `.shape` | Panjang pada setiap sumbu | `(3, 4)` |
| `.size` | Jumlah total elemen | `12` |
| `.dtype` | Tipe data umum dari elemen-elemen | `int64`, `float64`, dll. |

```python
measurement_matrix = np.array([[12, 15, 18, 21],
                               [14, 16, 20, 22],
                               [11, 13, 17, 19]])

print("Array:")
print(measurement_matrix)
print("ndim:", measurement_matrix.ndim)
print("shape:", measurement_matrix.shape)
print("size:", measurement_matrix.size)
print("dtype:", measurement_matrix.dtype)
```

### 3.2 Understanding Shape and Size

Bentuk `(3, 4)` berarti array memiliki tiga baris dan empat kolom; ukurannya (size) dengan demikian adalah `3 × 4 = 12` elemen. Untuk array 3D atau 4D, ukuran total adalah hasil kali seluruh angka pada bentuknya.

```python
print("1D:", scores_1d.ndim, scores_1d.shape, scores_1d.size)
print("2D:", scores_2d.ndim, scores_2d.shape, scores_2d.size)
print("3D:", observations_3d.ndim, observations_3d.shape, observations_3d.size)
print("4D:", observations_4d.ndim, observations_4d.shape, observations_4d.size)
```

### 3.3 Common NumPy Data Types (`dtype`)

NumPy memiliki tipe untuk bilangan bulat (`int`), bilangan desimal/titik mengambang (`float`), nilai Boolean (`bool`), dan teks (`str`). `dtype` aktual yang ditampilkan NumPy memberikan detail lebih lanjut, seperti jumlah bit yang digunakan untuk sebuah nilai numerik.

Karena array NumPy menggunakan satu tipe data yang sama, menggabungkan berbagai jenis nilai dapat menyebabkan konversi tipe otomatis.

```python
integer_values = np.array([1, 2, 3])
float_values = np.array([1.5, 2.5, 3.5])
boolean_values = np.array([True, False, True])
string_values = np.array(["rendah", "sedang", "tinggi"])

print("Bulat (int):", integer_values.dtype)
print("Desimal (float):", float_values.dtype)
print("Boolean:", boolean_values.dtype)
print("Teks (string):", string_values.dtype)
```

### 3.4 Specifying a Data Type Explicitly

Gunakan argumen `dtype` pada `np.array()` untuk meminta tipe tertentu. Tipe numerik yang lebih kecil dapat menempati memori lebih sedikit, tetapi rentang nilai atau presisinya juga lebih terbatas.

```python
measurements_float32 = np.array([1.25, 2.75, 3.50], dtype=np.float32)
counts_int32 = np.array([10, 20, 30], dtype=np.int32)

print(measurements_float32, measurements_float32.dtype)
print(counts_int32, counts_int32.dtype)
```

```python
measurements_int32 = np.array([1.25, 2.75, 3.50], dtype=np.int32)
measurements_int32 
```

### 3.5 Type Conversion with `astype()`

Metode `.astype()` mengembalikan array yang telah dikonversi ke tipe data yang ditentukan. Penerapan yang umum adalah mengubah string yang berisi angka valid menjadi nilai numerik.

Berhati-hatilah: mengonversi bilangan desimal ke bilangan bulat akan membuang bagian pecahannya; hal ini tidak sama dengan pembulatan ke bilangan bulat terdekat.

```python
text_scores = np.array(["70", "85", "90"])
numeric_scores = text_scores.astype(np.int64)

print("Sebelum:", text_scores, text_scores.dtype)
print("Sesudah:", numeric_scores, numeric_scores.dtype)
```

```python
continuous_values = np.array([1.9, 2.1, 3.8])
integer_conversion = continuous_values.astype(int)

print("Asli:", continuous_values)
print("Hasil konversi:", integer_conversion)
```

### 3.6 Type Coercion

**Type coercion** (pemaksaan tipe) terjadi ketika NumPy secara otomatis memilih satu tipe data yang sama untuk elemen-elemen yang digunakan membuat array. Contohnya, bilangan bulat diubah menjadi desimal ketika ada angka desimal yang disertakan, atau nilai numerik diubah menjadi string ketika dicampur dengan teks.

**Selalu periksa `.dtype` sebelum melakukan analisis numerik.**

```python
coercion_a = np.array([10, 20.5, 30])
coercion_b = np.array([True, False, 5])
coercion_c = np.array([1, "dua", 3])

print(coercion_a, coercion_a.dtype)
print(coercion_b, coercion_b.dtype)
print(coercion_c, coercion_c.dtype)
```

## 4. Reshaping Arrays

### 4.1 Understanding Array Shape

**Shape** menjelaskan bagaimana elemen-elemen disusun. Perlu diperhatikan bahwa array 1D dengan enam elemen memiliki bentuk `(6,)`, bukan `(1, 6)`. Array 2D berbentuk baris dan array 2D berbentuk kolom memiliki bentuk yang berbeda, meskipun berisi enam nilai yang sama.

```python
base_values = np.arange(1, 7)

row_array = base_values.reshape(1, 6)
column_array = base_values.reshape(6, 1)

print("Bentuk 1D:", base_values.shape)
print("Bentuk baris:", row_array.shape)
print(row_array)
print("Bentuk kolom:", column_array.shape)
print(column_array)
```

### 4.2 Reshaping Arrays with `reshape()`

`.reshape()` mengubah susunan elemen tanpa mengubah jumlah total elemen. Array 1D yang berisi 12 nilai dapat diubah bentuknya menjadi `(3, 4)`, `(4, 3)`, atau `(2, 2, 3)` karena setiap bentuk tersebut memiliki tepat 12 posisi.

Pada contoh dasar ini, NumPy membaca elemen sesuai urutan yang ada lalu menyusunnya ke dalam bentuk yang baru.

```python
values_12 = np.arange(1, 13)

matrix_3x4 = values_12.reshape(3, 4)
matrix_4x3 = values_12.reshape(4, 3)
array_2x2x3 = values_12.reshape(2, 2, 3)

print("Asli:", values_12)
print("3 × 4:")
print(matrix_3x4)
print("4 × 3:")
print(matrix_4x3)
print("2 × 2 × 3:")
print(array_2x2x3)
```

### 4.3 Reshape Compatibility

Jumlah total elemen **harus sesuai** dengan bentuk yang diminta. Misalnya, array berukuran 12 tidak dapat diubah menjadi `(5, 3)`, karena bentuk tersebut membutuhkan 15 elemen.

NumPy dapat menghitung **satu** dimensi yang hilang jika Anda menuliskan `-1`.

```python
reshaped_automatically = values_12.reshape(3, -1)

print(reshaped_automatically)
print("Bentuk hasil inferensi:", reshaped_automatically.shape)
```

### 4.4 Flattening Arrays with `flatten()`

`.flatten()` mengubah array multidimensi menjadi **salinan 1D yang baru**. Ini berguna ketika analisis membutuhkan seluruh elemen dalam satu urutan.

```python
original_matrix = np.array([[1, 2],
                            [5, 7],
                            [6, 6]])
flattened_values = original_matrix.flatten()

print("Asli:")
print(original_matrix)
print("Hasil flatten:")
print(flattened_values)
print("Bentuk hasil flatten:", flattened_values.shape)
```

### 4.5 `reshape()` vs. `flatten()`

| Operasi | Tujuan utama | Jumlah dimensi setelah operasi |
|---|---|---|
| `reshape(new_shape)` | Menyusun ulang elemen ke bentuk yang dipilih | Ditentukan oleh `new_shape` |
| `flatten()` | Mengubah seluruh elemen menjadi array 1D baru | Selalu 1D |

## 5. Basic Array Indexing and Slicing

### 5.1 Indexing a One-Dimensional Array (1D)

**Indexing** memilih sebuah elemen berdasarkan posisinya. Python menggunakan **indeks berbasis nol**, sehingga indeks `0` adalah elemen pertama, indeks `1` adalah elemen kedua, dan seterusnya.

Indeks negatif menghitung mundur dari akhir: `-1` adalah elemen terakhir.

```python
values = np.array([2, 4, 6, 8, 10])

print("Elemen pertama:", values[0])
print("Elemen keempat:", values[3])
print("Elemen terakhir:", values[-1])
```

### 5.2 Indexing a Two-Dimensional Array (2D)

Array 2D membutuhkan **indeks baris** dan **indeks kolom** untuk memilih satu elemen. Sintaksnya adalah `array[baris, kolom]`. Kedua indeks dimulai dari nol.

```python
marks = np.array([[80, 85, 90],
                  [75, 88, 84],
                  [92, 90, 95]])

print(marks)
print("Baris 1, kolom 2:", marks[1, 2])
print("Baris 2, kolom 0:", marks[2, 0])
```

### 5.3 Selecting Rows and Columns

Gunakan satu indeks untuk memilih sebuah baris (`array[baris]`) atau gabungkan indeks dengan `:` untuk memilih satu kolom utuh (`array[:, kolom]`). Tanda titik dua berarti **semua posisi** pada sumbu tersebut.

```python
print("Baris pertama:", marks[0])
print("Baris kedua:", marks[1, :])
print("Kolom ketiga:", marks[:, 2])
print("Kolom pertama:", marks[:, 0])
```

### 5.4 Slicing a One-Dimensional Array (1D)

**Slicing** mengambil beberapa elemen yang berurutan. Sintaks umumnya adalah `array[start:stop]`.

- `start` disertakan.
- `stop` tidak disertakan.
- `start` yang dihilangkan berarti awal array.
- `stop` yang dihilangkan berarti akhir array.

```python
sequence = np.array([10, 20, 30, 40, 50, 60])

print("Posisi 1 sampai 3:", sequence[1:4])
print("Tiga pertama:", sequence[:3])
print("Dari posisi 3 dan seterusnya:", sequence[3:])
print("Semua elemen:", sequence[:])
```

### 5.5 Slicing a Two-Dimensional Array (2D)

Slice 2D menentukan rentang baris dan rentang kolom: `array[baris_awal:baris_akhir, kolom_awal:kolom_akhir]`. Ini berguna untuk mengambil submatriks.

```python
number_grid = np.arange(1, 26).reshape(5, 5)

print("Matriks asli:")
print(number_grid)
print("Baris 1–3, kolom 2–4:")
print(number_grid[1:4, 2:5])
```

### 5.6 Slicing with Steps

Sebuah slice dapat menyertakan argumen ketiga: `array[start:stop:step]`. `step` menentukan jarak antarposisi yang dipilih.

- `::2` memilih setiap elemen kedua, mulai dari awal.
- `1::2` memilih setiap elemen kedua, mulai dari indeks 1.
- `[::-1]` membalik array 1D.

Step juga dapat diterapkan secara terpisah pada sumbu baris dan kolom dari array 2D.

```python
print("Setiap elemen kedua:", sequence[::2])
print("Setiap elemen kedua mulai indeks 1:", sequence[1::2])
print("Dibalik:", sequence[::-1])
print("Setiap baris dan kolom kedua:")
print(number_grid[::2, ::2])
```

### 5.8 Common Indexing and Slicing Mistakes

1. **Kesalahan selisih satu (off-by-one):** `a[1:4]` memilih indeks 1, 2, dan 3, bukan indeks 4.
2. **Tertukar urutan baris dan kolom:** `a[2, 1]` berarti indeks baris 2 dan indeks kolom 1.
3. **Keliru antara pemilihan elemen dan baris:** `a[1]` mengembalikan satu baris dari array 2D; `a[1, 1]` mengembalikan satu elemen.
4. **Indeks di luar jangkauan:** array dengan 5 elemen memiliki indeks positif yang valid dari 0 sampai 4.
5. **Keliru antara bentuk 1D dan 2D:** array 1D berukuran 5 memiliki bentuk `(5,)`, bukan `(1, 5)`.

**Perbedaan konseptual:** Indexing umumnya memilih satu posisi; slicing memilih rentang posisi. Pemilihan bersyarat dan modifikasi nilai array yang lebih lanjut akan dibahas pada Pertemuan 2.
